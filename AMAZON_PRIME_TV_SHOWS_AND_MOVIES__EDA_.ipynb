```python
# Analyse show type column
print('\n --Content type column--')
Show_type= Merged_df['type'].value_counts()
display(Show_type)

# Visualizing the Show_type by pie chart
print('\n--Pie Chart--')
plt.figure(figsize=(8, 6))
plt.pie(Show_type.values, labels=Show_type.index, autopct='%1.1f%%', startangle=140)
plt.title('Show Type Distribution')
plt.show()
```

```python
# Analyze the release year
print('\n --Release year--')
Release_year= Merged_df['release_year'].value_counts()
display(Release_year)

# Visualizing the release_year by histogram
print('\n--Histogram--')
plt.figure(figsize=(8, 6))
plt.hist(Merged_df['release_year'], bins=20, color='green', edgecolor='black')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.title('Release Year Distribution')
plt.show()
```

```python
# Analyse the runtime
print('\n --Runtime--')
Runtime= Merged_df['runtime'].describe()
display(Runtime)
# Separate runtime for TV shows and movies
TV_shows= Merged_df[Merged_df['type'] == 'SHOW']
Movies= Merged_df[Merged_df['type'] == 'MOVIE']

# Visualizing the separate runtime of TV shows and movies  by histogram
plt.figure(figsize=(12, 6))

# Histogram for TV shows
print('\n--Histogram of Tv shows--')
plt.subplot(1, 2, 1)
TV_shows['runtime'].hist(bins=30, color='skyblue', edgecolor='black')
plt.xlabel('Runtime (minutes)')
plt.ylabel('Number of TV shows')
plt.title('TV Show Runtime Distribution')

# Histogram for movies
print('\n--Histogram of Movies--')

plt.subplot(1, 2, 2)
Movies['runtime'].hist(bins=30, color='skyblue', edgecolor='black')
plt.xlabel('Runtime (minutes)')
plt.ylabel('Number of movies')
plt.title('Movie Runtime Distribution')


plt.tight_layout()
plt.show()
```

```python
# Analyze age certification
print('\n --Age certification--')
Age_certification= Merged_df['age_certification'].value_counts()
display(Age_certification)

# Visualize the age certification by bar chart
print('\n--Bar chart--')
plt.figure(figsize=(10, 6))
plt.bar(Age_certification.index, Age_certification.values, color='yellow', edgecolor='black')
plt.xlabel('Age Certification')
plt.ylabel('Number of Titles')
plt.title('Age Certification Distribution')
plt.xticks(rotation=45)
plt.show()
```

```python
import ast

# Analyze the genres
# convert the string representation of list t actual list
print("\n--Genres--")

Merged_df["genres_list"]= Merged_df['genres'].apply(ast.literal_eval)
# Flatten the list of genres and count occurances
all_genres =[genre for sublist in Merged_df['genres_list'] for genre in sublist]
genre_counts= pd.Series(all_genres).value_counts()
display(genre_counts)
# Visualise genres by horizontal bar chart
print('\n--Horizontal bar chart--')
plt.figure(figsize=(10, 6))
plt.barh(genre_counts.sort_values(ascending=True).index, genre_counts.sort_values(ascending=True).values, color='cyan', edgecolor='black')
plt.xlabel('Number of titles')
plt.ylabel('Genres')
plt.title('Genre Distribution')
plt.show()
```

The next chart we have is Horizontal bar chart showing the genres, there are different kind of genres which interest viewers and the most watched movies are drama movies showing 70000 movies.

```python
# Analyze production country
# convert the string representation of list t actual list
Merged_df["production_countries_list"]= Merged_df['production_countries'].apply(ast.literal_eval)
# Flatten the list of production_countries and count occurances
print('\n--Production countries--')
all_countries =[country for sublist in Merged_df['production_countries_list'] for country in sublist]
country_count= pd.Series(all_countries).value_counts()
display(country_count)

# Visualize the production_countries by horizontal bar graph
print('\n--Horizontal bar chart--')
plt.figure(figsize=(10, 6))
country_count.head(10).sort_values(ascending=True).plot(kind='barh', color='magenta', edgecolor='black')
plt.title('Top 10 Production Countries')
plt.xlabel('Number of titles')
plt.ylabel('Production Countries')
plt.show()
```

# Multivariate analysis

```python
import ast
# Get the top genres
all_genres = Merged_df['genres'].apply(ast.literal_eval).explode()
top_genres = all_genres.value_counts().head(10).index.tolist()

# Filter the DataFrame to include only the top genres
df_top_genres = Merged_df[Merged_df['genres'].apply(lambda x: any(genre in top_genres for genre in ast.literal_eval(x)))]

# Explode the genres column to have one row per genre and filter for top genres
df_top_genres_exploded = df_top_genres.assign(genres=df_top_genres['genres'].apply(ast.literal_eval)).explode('genres')
df_top_genres_exploded = df_top_genres_exploded[df_top_genres_exploded['genres'].isin(top_genres)]


# Calculate the average IMDB score for each type and genre combination
heatmap_data = df_top_genres_exploded.pivot_table(index='genres', columns='type', values='imdb_score', aggfunc='mean')

# Create the heatmap
plt.figure(figsize=(10, 7))
sns.heatmap(heatmap_data, annot=True, fmt=".2f", cmap="viridis")
plt.xlabel('Type')
plt.ylabel('Genre')
plt.title('Average IMDB Score by Type and Top Genres (Heatmap)')
plt.tight_layout()
plt.show()
```

Here i have analyzed top generes movies and shows with average imdb scores using heatmap.
* The yellow effect in rating shows average rating of 7.2 given to action genre and
*  average lower rating of 4 is given to horror genre.

```python

# Calculate the average tmdb popularity for each release year and show type combination
average_popularity_by_year_type = Merged_df.groupby(['release_year', 'type'])['tmdb_popularity'].mean().reset_index()

# Create the line chart
plt.figure(figsize=(12, 6))
sns.lineplot(data=average_popularity_by_year_type, x='release_year', y='tmdb_popularity', hue='type')
plt.xlabel('Release Year')
plt.ylabel('Average TMDB Popularity')
plt.title('Average TMDB Popularity by Release Year and Show Type')
plt.grid(True)
plt.show()
```

The last graph is line graph showing the popularity of movies and Tv shows  according to there release year. The blue line indicates movies and orange is Tv shows.
* As we can see in Tv shows between the 1940 to 1960 the graph has a peek popularity of 80 tp 90
